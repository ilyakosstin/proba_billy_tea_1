## Задача 3

$A_{i} = \{X = i\}$, $B_{i} = \{Y = i\}$ при $i \ge 0$

Для $\forall i \ge 0, j \ge 0$: $A_{i}$ и $B_{j}$ независимы

$$P(X = i) = e^{-\lambda} \frac{\lambda^{i}}{i!} \text{, } \lambda > 0 \text{, } i \ge 0$$
$$P(Y = j) = e^{-\mu} \frac{\mu^{j}}{j!} \text{, } \mu > 0 \text{, } j \ge 0$$

Найти: $P(X = i \mid X + Y = j)$

### Решение

По теореме Байеса

$P(X = i \mid X + Y = j) = \frac{P(X + Y = j \mid X = i) P(X = i)}{P(X + Y = j)} = \frac{P(Y = j - i) P(X = i)}{P(X + Y = j)}$

Найдем вероятность в знаменателе

$P(X + Y = j) = \sum_{k=0}^{j} P(X = k \& Y = j - k) = \sum_{k=0}^{j} P(X = k) P(Y = j - k) = \sum_{k=0}^{j} e^{-\lambda} \frac{\lambda^{k}}{k!} e^{-\mu} \frac{\mu^{j - k}}{(j - k)!} = e^{-\lambda - \mu} \sum_{k=0}^{j} \mu^{j} (\frac{\lambda}{\mu})^{k} \frac{1}{k!(j-k)!} = \frac{\mu^{j} e^{-\lambda - \mu}}{j!} \sum_{k=0}^{j} C_{j}^{k} (\frac{\lambda}{\mu})^{k} = \frac{\mu^{j} e^{-\lambda - \mu}}{j!} (\frac{\lambda}{\mu} + 1)^{j} = e^{-(\lambda + \mu)} \frac{(\lambda + \mu)^{j}}{j!}$

Подставим

$P(X = i \mid X + Y = j) = e^{-\mu} \frac{\mu^{j - i}}{(j - i)!} e^{-\lambda} \frac{\lambda^{i}}{i!} e^{\lambda + \mu} \frac{j!}{(\lambda + \mu)^{j}} = C_{j}^{i} (\frac{\lambda}{\mu})^{i} (\frac{\mu}{\lambda + \mu})^{j}$

Учтем также ограничения на $i$.

Ответ:

$$P(X = i \mid X + Y = j) = \begin{cases} C_{j}^{i} \left(\frac{\lambda}{\mu}\right)^{i} \left(\frac{\mu}{\lambda + \mu}\right)^{j}, & \text{if } i \in [0:j] \\ 0, & \text{if } i \notin [0:j] \end{cases}$$
